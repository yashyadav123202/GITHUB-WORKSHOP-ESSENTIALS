*Combining and Merging Datasets*

```python
class_a = pd.DataFrame({
    "Student_ID":[1,2],
    "Name":["Yuvraj","Ujjwal"],
    "Marks":["87","91"]
})
class_b = pd.DataFrame({
    "Student_ID":[3,4],
    "Name":["Yuvraj","Ujjwal"],
    "Marks":["87","91"]
})
all_students = pd.concat([class_a,class_b],ignore_index=True)
print(all_students)
```

```python
all_students = pd.concat([class_a,class_b],axis=1)
print(all_students)
```

```python
all_students = pd.concat([class_a,class_b],axis=0)
print(all_students)
```

Merge

```python
student = pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Yuvraj","Ujjwal","Aman","Riya"],
})
result = pd.DataFrame({
    "Student_ID":[1,2,3,5],
    "Marks":["87","91","85","92"]
})
```

```python
merged = pd.merge(student,result,on = "Student_ID",how = "left")
print(merged)
```

```python
merged = pd.merge(student,result,on = "Student_ID",how = "right")
print(merged)
```

```python
merged = pd.merge(student,result,how = "cross")
print(merged)
```

Merging when key-column names differ

```python
student_info = pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Yuvraj","Ujjwal","Aman","Riya"],
})
result_info = pd.DataFrame({
    "ID":[1,2,3,5],
    "Marks":["87","91","85","92"]
})
answer = pd.merge(student_info,result_info,left_on = "Student_ID",right_on = "ID",how = "left")
print(answer)
```

Reshaping and Pivoting

```python
wide = pd.DataFrame({
    "Name":["Riya","Aman","Yuvraj","Ujjwal"],
    "Maths":[87,85,88,90],
    "Science":[91,98,95,67],
    "English":[82,89,84,86]
})
print(wide)
```

Convert wide data to long using melt()

```python
long = pd.melt(wide,id_vars = ["Name"],var_name = "Subject",value_name = "Marks")
print(long)
```

convert long data to wide using pivot()

```python
wide_again = long.pivot(
    index = "Name",
    columns="Subject",
    values="Marks"
)
print(wide_again)
```

```python
sales =  pd.DataFrame({
    "City":["Agra","Agra","Delhi","Delhi","Agra"],
    "Product":["Pen","Pen","Pen","Pencil","Pencil"],
    "Sales":[100,200,150,300,250]
})
summary = sales.pivot_table(
    index = "City",
    columns = "Product",
    values = "Sales",
    aggfunc = "sum",
    fill_value=0
)
print(summary)

```

Crosstabulation

```python
data = pd.DataFrame({
    "Dept":["CSE","CSE","ECE","ECE","CSE"],
    "Result":["Pass","Fail","Pass","Pass","Pass"],
})
table = pd.crosstab(data["Dept"],data["Result"])
print(table)
```
