```python
# Q0. Upload the required CSV files in Google Colab
from google.colab import files
uploaded = files.upload()

import pandas as pd
import numpy as np
from IPython.display import display

# Read every uploaded CSV using its filename as the key
data = {name: pd.read_csv(name) for name in uploaded if name.lower().endswith(".csv")}

def load_csv(filename):
    if filename not in data:
        raise FileNotFoundError(
            f"{filename} was not uploaded. Upload all CSV files listed in the practice sheet."
        )
    return data[filename].copy()

# Load commonly used files
students = load_csv("students_master.csv")
exam = load_csv("exam_marks.csv")
ca = load_csv("section_CA.csv")
cb = load_csv("section_CB.csv")
original = load_csv("marks_original.csv")
updated = load_csv("marks_updated.csv")
long_marks = load_csv("marks_long.csv")
```

## Q12. Use combine() with np.maximum

```python
maximum_marks = original_idx.combine(
    updated_idx,
    lambda s1, s2: np.maximum(s1, s2)
)
display(maximum_marks)
```

**Observation:** `np.maximum` propagates a missing value (`NaN`) if either corresponding value is missing.

## Q13. Use combine() with np.fmax and compare with Q12

```python
fmax_marks = original_idx.combine(
    updated_idx,
    lambda s1, s2: np.fmax(s1, s2)
)
display(fmax_marks)

comparison_q12_q13 = pd.concat(
    [maximum_marks.rename(columns={"Marks": "np.maximum"}),
     fmax_marks.rename(columns={"Marks": "np.fmax"})],
    axis=1
)
display(comparison_q12_q13)
```

**Observation:** `np.fmax` returns the non-missing value when only one value is missing. `np.maximum` returns `NaN` in that case.

## Q24. Attempt pivot() on repeat_tests.csv and explain the issue

```python
repeat_tests = load_csv("repeat_tests.csv")

try:
    repeat_pivot = repeat_tests.pivot(
        index="Student_ID", columns="Subject", values="Marks"
    )
    display(repeat_pivot)
except ValueError as error:
    print("Pivot failed:", error)
```

**Explanation:** `pivot()` requires each `(Student_ID, Subject)` pair to identify exactly one value. Repeated tests create duplicate pairs, so Pandas cannot decide which mark should occupy the cell.

## Q27. Combine marks by Student_ID and merge with all students

```python
combined_by_id = original_by_id.combine_first(updated_by_id)

all_students_marks = pd.merge(
    students,
    combined_by_id.reset_index(),
    on="Student_ID",
    how="left"
)
display(all_students_marks)

print("Final source check for Student_ID 101 and 102:")
for sid in [101, 102]:
    original_value = original_by_id["Marks"].get(sid, np.nan)
    updated_value = updated_by_id["Marks"].get(sid, np.nan)
    final_value = combined_by_id["Marks"].get(sid, np.nan)
    source = "original file" if pd.notna(original_value) else (
        "updated file" if pd.notna(updated_value) else "neither file"
    )
    print(f"Student_ID {sid}: original={original_value}, "
          f"updated={updated_value}, final={final_value}, source={source}")
```

**Observation:** `combine_first()` takes the original mark when it is present; otherwise it fills the missing value from the updated file. The printed check identifies the source for IDs 101 and 102 using the actual CSV data.

## Q28. Compare combine_first() and combine(np.fmax) side by side

```python
result_combine_first = original_by_id.combine_first(updated_by_id)
result_fmax = original_by_id.combine(
    updated_by_id,
    lambda s1, s2: np.fmax(s1, s2)
)

side_by_side = pd.concat(
    [
        result_combine_first.rename(columns={"Marks": "combine_first"}),
        result_fmax.rename(columns={"Marks": "np.fmax"})
    ],
    axis=1
).sort_index()

display(side_by_side)
different_ids = side_by_side[
    side_by_side["combine_first"].ne(side_by_side["np.fmax"]) |
    (side_by_side["combine_first"].isna() != side_by_side["np.fmax"].isna())
]
print("Student_IDs where results differ:")
display(different_ids)
```

**Observation:** `combine_first()` prefers the original value whenever it exists. `np.fmax` selects the larger available value. They differ when both values exist and the updated mark is higher (and can also differ in missing-value handling in other cases).

## Q29. Create mean and maximum pivot table, stack/unstack, and restore layout

```python
repeat_stats = repeat_tests.pivot_table(
    index="Student_ID",
    columns="Subject",
    values="Marks",
    aggfunc=["mean", "max"]
)
print("Pivot table with mean and max:")
display(repeat_stats)

# Stack the aggregation-function column level
stacked_stats = repeat_stats.stack(level=0)
print("After stack():")
display(stacked_stats)

# Unstack to restore the wide layout, then reorder column levels
restored_stats = stacked_stats.unstack(level=-1)
restored_stats = restored_stats.swaplevel(0, 1, axis=1).sort_index(axis=1)

print("Restored pivot-table layout:")
display(restored_stats)
print("Matches original pivot table:", restored_stats.equals(repeat_stats))
```

## Q30. Merge students with long marks, pivot to wide format, and reset_index

```python
students_long = pd.merge(
    students, long_marks, on="Student_ID", how="inner"
)

wide_with_name = students_long.pivot(
    index=["Student_ID", "Student_Name"],
    columns="Subject",
    values="Marks"
)

wide_reset = wide_with_name.reset_index()
display(wide_reset)

print("Columns after reset_index:", wide_reset.columns.tolist())
```

**Explanation:** This `pivot()` succeeds when each `(Student_ID, Student_Name, Subject)` combination occurs only once in `marks_long.csv`. The `repeat_tests.csv` data has repeated `(Student_ID, Subject)` combinations, so `pivot()` fails there. If the long marks file also contains duplicate combinations, this pivot will fail for the same reason.
